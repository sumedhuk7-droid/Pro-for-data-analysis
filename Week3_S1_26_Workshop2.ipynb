**<h1><center> Broadcasting </center></h1>**

```python
a = np.array([2.0, 4.0, 6.0])
b = np.array([7.0, 8.0, 9.0, 10.0])
print(a,"  ",b)
```

```python
a+b

# Error? Understand why...
```

**Solutions:**

```python
a = np.array([2.0, 4.0, 6.0])
b = np.array([7.0, 8.0, 9.0, 10.0])

# Find the difference in array lengths
difference = len(b) - len(a)

# Add zeros to the shorter array
a = np.pad(a, (0, difference), mode='constant')

# Add the arrays
c = a + b

print(c)
```
